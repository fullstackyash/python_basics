# Lambda Functions

### Lambda functions are one liner functions defined by lambda keyword

# Syntax of Lambda Function

### Lambda Arguments:Expressions

### Lambda --> Keyword which tells python interpreter know that the current function is a lambda function

### Arguments --> Input values for our function 

### Expression --> Logic (Function Body)

### 1) Lambda Functions can have any number of arguments but only on logic or expression
### 2) They are used for shor operations or arguments to Higher order functions
### 3) They can be treated as a function without a name & they do not even require a return statement

### example 
### lambda x:x**2 // square of a number

```python
def square(number):
    return number**2
```

```python
print(square(5))  # Output: 25
```

```python
## With lambda function
square_lambda = lambda number:number**2
print(square_lambda(5))  # Output: 25
type(square_lambda)  # Output: <class 'function'>
```

```python
##  lambda function for sum of two numbers
sum_lambda = lambda a, b: a + b
print(sum_lambda(3, 4))  # Output: 7
```

```python
sum_lambda(10, 20)  # Output: 30   
```
