### Python standard library is a vast collection of modules and packages, that come bundled with python and it provides us with wide range of functionalities.

```python
# Math Library -> it provides mathematical functions and constants
import math 
# Example usage of math library
print("Square root of 16 is:", math.sqrt(16))
print("Value of pi is:", math.pi)
```

```python
# Random Library -> it provides functions for generating random numbers and performing random operations
import random
# Example usage of random library
print("Random integer between 1 and 10:", random.randint(1, 10))
print("Random choice from a list:", random.choice(['apple', 'banana', 'cherry']))
```

```python
# CSV Library -> it provides functions for reading and writing CSV (Comma Separated Values) files
import csv
# Example usage of csv library
with open('example.csv', mode='w', newline='') as file:
    writer = csv.writer(file)
    writer.writerow(['Name', 'Age', 'City'])
    writer.writerow(['Alice', 30, 'New York'])
    writer.writerow(['Bob', 25, 'Los Angeles'])
print("CSV file 'example.csv' has been created with sample data.")
```

```python
with open('example.csv', mode='r') as file:
    reader = csv.reader(file)
    for row in reader:
        print(row)
```
