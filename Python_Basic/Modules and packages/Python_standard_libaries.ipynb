### Python standard library is a vast collection of modules and packages, that come bundled with python and it provides us with wide range of functionalities.

```python
# Math Library -> it provides mathematical functions and constants
import math 
# Example usage of math library
print("Square root of 16 is:", math.sqrt(16))
print("Value of pi is:", math.pi)
```

```python
# Random Library -> it provides functions for generating random numbers and performing random operations
import random
# Example usage of random library
print("Random integer between 1 and 10:", random.randint(1, 10))
print("Random choice from a list:", random.choice(['apple', 'banana', 'cherry']))
```

```python
# CSV Library -> it provides functions for reading and writing CSV (Comma Separated Values) files
import csv
# Example usage of csv library
with open('example.csv', mode='w', newline='') as file:
    writer = csv.writer(file)
    writer.writerow(['Name', 'Age', 'City'])
    writer.writerow(['Alice', 30, 'New York'])
    writer.writerow(['Bob', 25, 'Los Angeles'])
print("CSV file 'example.csv' has been created with sample data.")
```

```python
with open('example.csv', mode='r') as file:
    reader = csv.reader(file)
    for row in reader:
        print(row)
```

```python
# Datetime Library -> it provides functions for working with dates and times
import datetime
# Example usage of datetime library
current_date = datetime.date.today()
print("Current date is:", current_date)
current_time = datetime.datetime.now().time()
print("Current time is:", current_time)
```

```python
timedelta_example = datetime.timedelta(days=5)
print("Date after 5 days will be:", current_date + timedelta_example)
```

```python
# Time Library -> it provides functions for working with time-related tasks, such as measuring time intervals and pausing execution and current time
import time
# Example usage of time library
print("Current time in seconds since the epoch:", time.time())
print("Pausing execution for 2 seconds...")
time.sleep(2)
print("Execution resumed after 2 seconds.")
```

```python
# Os Library -> it provides functions for interacting with the operating system, such as file and directory operations
import os
# Example usage of os library
print("Current working directory:", os.getcwd())
os.mkdir('new_directory')
print("Directory 'new_directory' has been created.")
```
