```python
## Exception Handling in Python
# Exception handling is a crucial aspect of programming that allows developers to manage and respond to errors gracefully. In Python, exceptions are handled using the `try`, `except`, `else`, and `finally` blocks. This structure helps ensure that your program can continue running even when unexpected issues arise.
```

```python
# Types of Errors in Exception Handling 
# ZeroDivisionError: This occurs when you attempt to divide a number by zero.
# FileNotFoundError: This occurs when you try to access a file that does not exist.
# ValueError: This occurs when a function receives an argument of the correct type but an inappropriate value.
# TypeError: This occurs when an operation or function is applied to an object of inappropriate type.
# KeyError: This occurs when you try to access a key that does not exist in a dictionary.
# IndexError: This occurs when you try to access an index that is out of range in a list or tuple.  

```

```python
# Types of Exception Handling
# 1. Try-Except Block: This is the most basic form of exception handling. You wrap the code that may raise an exception in a `try` block, and then handle the exception in the `except` block.
try:
    result = 10 / 0
except ZeroDivisionError:
    print("You cannot divide by zero!")
```

```python
# If you want to handle multiple exceptions, you can specify them in separate `except` blocks.
denominator = input("Enter a number to divide 10 by: ")
try:
    value = 10/int(denominator)
    print("Result:", value)
except ZeroDivisionError:
    print("You cannot divide by zero!")

except TypeError as t:
    print("Type error occurred:", t)
except ValueError as v:
    print("Value error occurred:", v)    
```
