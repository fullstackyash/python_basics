# OS Library

### OS Library helps us to interact with our operating system, an lets us work with multiple files and directories.

### getcwd() -> provides us the path of the current working Directory

```python
import os
```

```python
current_working_dir = os.getcwd()
print(f"Current working directory: {current_working_dir}")
```

### mkdir() - Create a new directory

```python
new_directory = "new_folder"
os.mkdir(new_directory)
print(f"Directory '{new_directory}' created.")
```

### listdir() -> Lists all the directories

```python
all_directories = os.listdir('.')
print(all_directories)
```

### os.path.join(current_working_dir, new_directory) -> Joins the current working directory with the new directory to create a full path.


```python
new_directory = "new_folder"
new_file = "new_file.txt"
new_file_path = os.path.join(new_directory, new_file)
print(f"New file path: {new_file_path}")
```

```python
print(os.path.join(os.getcwd(), new_directory, new_file))
```

### os.path.exists() -> Checks if a file or directory exists

```python
print(os.path.exists(new_directory))
```

```python
print(os.path.exists(new_file))
```

### os.path.isdir() -> Checks if the path is a directory
### os.path.isfile() -> Checks if the path is a file

```python
if(os.path.isfile(new_directory)):
    print(f"{new_file_path} is a file.")
elif(os.path.isdir(new_directory)):
    print(f"{new_file_path} is a directory.")
else:
    print(f"{new_directory} is not a file or dir .")
```

```python
### os.path.abspath() -> Returns the complete path or absolute path of a file or directory
```

```python
relative_path = "new_folder"
absolute_path = os.path.abspath(relative_path)
print(f"Absolute path of '{relative_path}': {absolute_path}")
```

```python
relative_path = "new_file.txt"
absolute_path = os.path.abspath(relative_path)
print(f"Absolute path of '{relative_path}': {absolute_path}")
```
