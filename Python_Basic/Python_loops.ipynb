# Print even numbers from 0 to 50 

## For Loop



```python
for i in range(51):
    if i%2==0:
        print(i)
```

## While Loop

```python
i=0
while i<50:
    if i%2==0:
        print(i)
    i+=1
```

# Break and Continue

```python
## Break terminates the loop when a certain condition is met
for i in range(10):
    if i==5:
        break
    print(i)
```

```python
## continue skips the current iteration and moves to the next one
for i in range(10):
    if i==5:
        continue
    print(i)
```

```python
# Prime number from 1 to 100
for num in range(1, 101):
    if num > 1:  # 1 is not a prime number
        for i in range(2, int(num**0.5) + 1):
            if num % i == 0:
                break
        else:
            print(num)
```
