# Filter function will retrieve  all the values from an iterable that meet a certain condition

# Filter function creates an iterator from elements in an iterable when a function returns true

# Syntax of filter function

# list(filter(function,iterable))

```python
# filter values to retrieve only even numbers

def is_even(x):
    return x % 2 == 0
```

```python
list_filter = filter(is_even, [1, 2, 3, 4, 5])
print(list(list_filter))
```

```python
#solving same problem with lambda function
filter_lambda = filter(lambda x:x%2==0, [1, 2, 3, 4, 5])
print(list(filter_lambda))
```
