# File Handling

### Data can be stored in Ram or a file
 
### In Ram data is stored temporally   but in file data is stored permanently  

# Types of file handling operation 

### Open --In order to read or write a file we need to first open the file

### Open(fileName, mode) function is used to open a file
### 1st parameter is the name of the file, 2nd parameter is the mode in which we want to open the file

### Type of mode
### r -> Read mode, to just read the contents of the file
### w -> Write mode the just write the contents of the file
### r+ -> To first read the file and then write in the file
### w+ -> To write the file first and then read the file
