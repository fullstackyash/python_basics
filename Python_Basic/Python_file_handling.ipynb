# File Handling

### Data can be stored in Ram or a file
 
### In Ram data is stored temporally   but in file data is stored permanently  

# Types of file handling operation 

### Open --In order to read or write a file we need to first open the file

### Open(fileName, mode) function is used to open a file
### 1st parameter is the name of the file, 2nd parameter is the mode in which we want to open the file

### Type of mode
### r -> Read mode, to just read the contents of the file
### w -> Write mode the just write the contents of the file
### r+ -> To first read the file and then write in the file
### w+ -> To write the file first and then read the file

```python
#Read example 
file=open('example.txt', 'r')
content=file.read()
print(content)
file.close()
```

### Write operation can be done in 2 ways
### Mode -> w, This will overwrite the contents and add new content into the file
### Mode -> a,  This function will append new content to the existing content at the end in the file

```python
# Example of write operation
file=open('example.txt', 'w')
file.write('This is Yash, I am a Fullstack AI Engineer, Having 8+ years of experience')
file.close()
```

```python
# example of append operation
file=open('example.txt', 'a')
file.write('\nI am also a Data Scientist, I have worked on 100+ projects')
file.close()
```

### With -> Statement use as a context manager in python with allows files to be used as objects and insures safe closing of file without using the traditional try and finally blocks and using close() function

```python
with open('example.txt', 'r') as file:
    content=file.read()
    print(content)
```
