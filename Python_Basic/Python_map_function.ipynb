# map()

# Map function applies a function or operation to all the items of the iterable and returns a map object 

# map function is useful to transform the data or applying an operation to all the items of an iterable

# map() helps us avoid unnecessary  lines of code  

# Syntax of Map()

# list(map(function,iterable))

```python
#squaring all the numbers in a list
# items = [1, 2, 3, 4, 5]
def square(list_items):
    empty_list = []
    for items in list_items:
        empty_list.append(items**2)
    return empty_list
```

```python
square([1, 2, 3, 4, 5])
```
