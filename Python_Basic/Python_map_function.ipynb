# map()

# Map function applies a function or operation to all the items of the iterable and returns a map object 

# map function is useful to transform the data or applying an operation to all the items of an iterable

# map() helps us avoid unnecessary  lines of code  

# Syntax of Map()

# list(map(function,iterable))

```python
#squaring all the numbers in a list
# items = [1, 2, 3, 4, 5]
def square(list_items):
    empty_list = []
    for items in list_items:
        empty_list.append(items**2)
    return empty_list
```

```python
square([1, 2, 3, 4, 5])
```

```python
# similar problem with map function
list_map = map(lambda x:x**2, [1, 2, 3, 4, 5])
print(list(list_map))
```

```python
def mySquare(x):
    return x**2
our_function = map(mySquare, [1, 2, 3, 4, 5])
print(list(our_function))
```

```python
# map function with multiple iterables
numbers1 = [1, 2, 3]
numbers2 = [4, 5, 6]
result = map(lambda x, y: x + y, numbers1, numbers2)
print(list(result))
```

```python
# map to convert string to int
map_result = map(int, ['1', '2', '3'])
print(list(map_result))
```

```python
# using map in dictionary inside list
employees = [
    {'name': 'Alice', 'age': 30},
    {'name': 'Bob', 'age': 25},
    {'name': 'Charlie', 'age': 35}
]

def get_employee_names(employee):  
    return employee['name']

employee_names = map(get_employee_names, employees)
print(list(employee_names))

employee_ages = map(lambda emp: emp['age'], employees)
print(list(employee_ages))
```
