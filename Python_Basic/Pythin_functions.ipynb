# Function with Multiple parameters

```python
def addition(num1, num2):
    print(f'the addition of {num1} and {num2} is {num1+num2}')
```

```python
addition(5, 10)
```

# Function with Variable length arguments


```python
def addition_variable(*nums):
    add = 0
    for num in nums:
        add += num
    print(f'the addition of {nums} is {add}')
```

```python
addition_variable(10,20,30,40)
```

# Keyword Arguments -  variable length arguments in key value pair

```python
def user_info(**user_data):
    for key,value in user_data.items():
        print(f'{key} : {value}')
```

```python
user_info(name='Ajay', age=30, city='New York')
```


# Positional Arguments with Keyword Arguments

```python
def data(*positional, **keyword):
    for item in positional:
        print("Positional Arguments:", item)
    for key, value in keyword.items():
        print(f"Keyword Argument - {key}: {value}")
```

```python
data(1, 2, 3, name='Alice', age=25)
```
